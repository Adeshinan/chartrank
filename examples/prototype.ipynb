# ChartRank prototype

Task 1.3: see the proposed charts for **14 pairs** of columns on three datasets (Titanic, Palmer penguins, a small sales series).

Each `suggest()` call returns a gallery: the recommended chart (★) comes first, with a gold border. Prototype: Plotly default colours, no guard rules and no explanations yet.

```python
from pathlib import Path

import numpy as np
import pandas as pd

import chartrank as cr

DATA = Path("..") / "tests" / "data"
titanic = pd.read_csv(DATA / "titanic.csv")
penguins = pd.read_csv(DATA / "penguins.csv")

# A small daily sales series (deterministic, no random draw)
days = pd.date_range("2026-01-01", periods=180, freq="D")
t = np.arange(len(days))
sales = pd.DataFrame(
    {
        "day": days,
        "revenue": 200 + 0.8 * t + 25 * np.sin(2 * np.pi * t / 7),
        "channel": np.array(["web", "shop", "phone", "web", "web"])[t % 5],
    }
)
cr.profile(titanic)
```

## Titanic

**Pclass × Fare**, intent `compare`. Fare by class: the reference case.

```python
g = cr.suggest(titanic, x="Pclass", y="Fare", intent="compare")
g.show()
```

**Sex × Age**, intent `distribution`. Age distribution for women and men.

```python
g = cr.suggest(titanic, x="Sex", y="Age", intent="distribution")
g.show()
```

**Age × Fare**, intent `relationship`. Two numeric columns.

```python
g = cr.suggest(titanic, x="Age", y="Fare", intent="relationship")
g.show()
```

**Sex × Embarked**, intent `composition`. Two categories.

```python
g = cr.suggest(titanic, x="Sex", y="Embarked", intent="composition")
g.show()
```

**Survived × Fare**, intent `compare`. A boolean on x is drawn as a category.

```python
g = cr.suggest(titanic, x="Survived", y="Fare", intent="compare")
g.show()
```

**Fare × Survived**, intent `compare`. Numeric x boolean has its own row.

```python
g = cr.suggest(titanic, x="Fare", y="Survived", intent="compare")
g.show()
```

**Pclass × Survived**, intent `compare`. Class x survival: two categories.

```python
g = cr.suggest(titanic, x="Pclass", y="Survived", intent="compare")
g.show()
```

## Palmer penguins

**species × body_mass_g**, intent `compare`. Body mass by species.

```python
g = cr.suggest(penguins, x="species", y="body_mass_g", intent="compare")
g.show()
```

**bill_length_mm × bill_depth_mm**, intent `relationship`. Bill shape.

```python
g = cr.suggest(penguins, x="bill_length_mm", y="bill_depth_mm", intent="relationship")
g.show()
```

**island × species**, intent `composition`. Which species on which island?

```python
g = cr.suggest(penguins, x="island", y="species", intent="composition")
g.show()
```

**flipper_length_mm × sex**, intent `compare`. x and y are swapped.

```python
g = cr.suggest(penguins, x="flipper_length_mm", y="sex", intent="compare")
g.show()
```

**species × flipper_length_mm**, intent `distribution`. Flipper length.

```python
g = cr.suggest(penguins, x="species", y="flipper_length_mm", intent="distribution")
g.show()
```

## Sales (dates)

**day × revenue**, intent `trend`. Daily revenue over six months.

```python
g = cr.suggest(sales, x="day", y="revenue", intent="trend")
g.show()
```

**day × channel**, intent `composition`. Share of each sales channel.

```python
g = cr.suggest(sales, x="day", y="channel", intent="composition")
g.show()
```

## Using one chart

`g.best` is the ★ chart; `.figure` is a real Plotly figure that you can modify.

```python
g = cr.suggest(titanic, x="Pclass", y="Fare", intent="compare")
print(g)
fig = g.best.figure
fig.update_layout(title="Ticket price by class")
fig
```

## Clear errors

An impossible intent lists the possible ones.

```python
try:
    cr.suggest(titanic, x="Pclass", y="Fare", intent="trend")
except ValueError as error:
    print(error)
```
